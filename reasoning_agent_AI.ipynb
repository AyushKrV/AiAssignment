# Classical Reasoning System: entailment over rules, causal graphs and logic stories

**Task.** Given a knowledge base (context) and a query (statement), decide whether the query is entailed: `True` / `False`.

**Pipeline.** text → router → translator → (facts, rules, interventions, goal) → **forward chaining** → True/False

One classical reasoning engine serves RuleTaker, CounterBench and FOLIO; only the translator in front of it differs.
The translators are regular-expression templates. No generative model is used, and the decision always comes from the engine.

**To generate `submission.csv`:** set `TEST_PATH` in the last cell and run it. The sections before it (data loading and experiments) are optional for that step.

## 2. Reasoning engine: forward chaining to a fixpoint
* **atom**: a hashable name for a proposition; **literal** = (atom, polarity); **rule** = (list of body literals, head atom).
* **interventions** (`fixed`): atoms forced to True/False from outside, used for counterfactual questions.
* A negative literal holds when its atom is not derived (negation as failure).
* Each pass recomputes the derived set from the facts, so negation stays correct on acyclic graphs; for positive-only rules it is the usual least fixpoint.

```python
import re, time, csv
from collections import Counter

# literal = (atom, positive?)   rule = ([literals], head_atom)
NEG_MODE = "naf"        # "naf" (current) | "strict"

def run_engine(facts, rules, fixed=None):
    fixed = fixed or {}
    forced_true = {a for a, v in fixed.items() if v}
    forced_false = {a for a, v in fixed.items() if not v}
    heads = {h for _, h in rules}
    base = (set(facts) - forced_false) | forced_true
    known, why = set(base), {}

    def holds(atom, positive):
        if positive:
            return atom in known
        if NEG_MODE == "strict":        # absence counts only if it is set from outside
            return atom in forced_false or (atom not in known and atom not in heads)
        return atom not in known

    for _ in range(len(rules) + 2):
        new = set(base)
        for body, head in rules:
            if head in fixed:
                continue
            if all(holds(a, p) for a, p in body):
                new.add(head)
                why.setdefault(head, body)
        if new == known:
            break
        known = new
    return known, why

def explain(atom, why, seen=None, depth=0):
    """Print the derivation chain for an atom (for the report's reasoning traces)."""
    seen = seen if seen is not None else set()
    if atom in seen or atom not in why:
        print("  " * depth + f"{atom}  [given]")
        return
    seen.add(atom)
    print("  " * depth + f"{atom}  <-  {why[atom]}")
    for a, positive in why[atom]:
        if positive:
            explain(a, why, seen, depth + 1)
```

## 3. Translators (natural language → facts and rules)
### 3.1 RuleTaker
Facts are the lines under `Facts:`. Each rule line (five templates) becomes one rule `A -> B`. Atoms are normalised sentences.

```python
RT_PATTERNS = [
    re.compile(r"^assuming (.+?), we can say (.+)$"),
    re.compile(r"^if (.+?), then (.+)$"),
    re.compile(r"^based on (.+?), it is evident that (.+)$"),
    re.compile(r"^given that (.+?), it follows that (.+)$"),
    re.compile(r"^when (.+?), we can conclude (.+)$"),
]

def norm(text):
    text = text.lower().strip().rstrip(".")
    text = re.sub(r"\bthe\s+", "", text)
    return " ".join(text.split())

def frontend_ruletaker(context, statement):
    """RuleTaker text -> (facts, rules, no interventions, goal atom)."""
    facts, rules, section = set(), [], None
    for line in context.split("\n"):
        line = line.strip()
        if line == "":
            continue
        if line == "Facts:":
            section = "facts"; continue
        if line == "Rules:":
            section = "rules"; continue
        if section == "facts":
            facts.add(norm(line))
        elif section == "rules":
            low = line.lower().rstrip(".")
            for pattern in RT_PATTERNS:
                m = pattern.match(low)
                if m:
                    rules.append(([(norm(m.group(1)), True)], norm(m.group(2))))
                    break
    return facts, rules, {}, norm(statement)
```

### 3.2 CounterBench
`A causes B` → `A -> B`; `A and B together cause C` → one rule with two body literals; `A or B causes C` → two rules;
`A causes not B` → rule with a negated body. Variables that are never an effect are root facts unless observed false.
The question yields the target atom and the interventions (`not X` forces X to False). Nested questions apply both interventions (`NESTED_MODE = "both"`).

```python
def _lit(text):
    t = re.sub(r"^and\s+", "", text.strip().strip("."))
    neg = t.lower().startswith("not ")
    return (t[4:].strip() if neg else t.strip()), (not neg)

def _body_dnf(text):
    text = re.sub(r"\btogether\b", "", text).strip()
    text = re.sub(r"^and\s+", "", text)
    return [[_lit(a) for a in d.split(" and ")] for d in text.split(" or ")]

def _negate_dnf(dnf):
    result = [[]]
    for conj in dnf:                                  # not(C1 or C2) = not C1 and not C2
        options = [[(a, not p)] for a, p in conj]     # not(a and b) = not a or not b
        result = [r + o for r in result for o in options]
    return result

NESTED_MODE = "both"        # "both" | "first_only" | "last_only"  (compared in section 5)

def frontend_counterbench(context, statement):
    """CounterBench text -> (root facts, causal rules, interventions, target atom)."""
    text, observed = context.strip(), {}
    m = re.search(r"We observed (.+)$", text, re.S)
    if m:
        for part in re.split(r",\s*|\s+and\s+", m.group(1).strip(" .")):
            if part.strip():
                name, val = _lit(part)
                observed[name] = val
        text = text[:m.start()]
    text = re.sub(r"^\s*We know that\s*", "", text).strip().rstrip(". ")

    segments = re.split(r"\bcauses?\b", text)
    bodies, spec = [_body_dnf(segments[0])], []
    for k in range(1, len(segments)):
        pieces = [p.strip() for p in segments[k].split(",") if p.strip()]
        last = (k == len(segments) - 1)
        eff_pieces = pieces if last else pieces[:-1]
        effects = []
        for p in eff_pieces:
            for q in re.sub(r"^and\s+", "", p).split(" and "):
                effects.append(_lit(q))
        spec.append((bodies[-1], effects))
        if not last:
            bodies.append(_body_dnf(pieces[-1]))

    rules, heads, used = [], set(), set()
    for dnf, effects in spec:
        for name, positive in effects:
            for conj in (dnf if positive else _negate_dnf(dnf)):   # "causes not B": B = not body
                rules.append((conj, name))
                heads.add(name)
                used.update(a for a, _ in conj)
    roots = (used - heads) | {n for n in observed if n not in heads}
    facts = {r for r in roots if observed.get(r, True)}           # roots occur unless observed false

    q = statement.strip()
    mt = re.search(r"[Ww]ould\s+(\w+)\s+occur", q)
    if not mt:
        return facts, rules, {}, None
    target = mt.group(1)
    cond = re.split(r"\binstead of\b", q[:mt.start()] + " " + q[mt.end():])[0]
    atoms = heads | used | {target}
    ordered = [(m2.group(2), m2.group(1) is None)
               for m2 in re.finditer(r"(not\s+)?\b([A-Z]\w*)\b", cond) if m2.group(2) in atoms]
    if re.search(r"further suppose|based on this assumption", q, re.I) and len(ordered) > 1:
        if NESTED_MODE == "first_only":
            ordered = ordered[:1]
        elif NESTED_MODE == "last_only":
            ordered = ordered[-1:]
    return facts, rules, dict(ordered), target
```

### 3.3 FOLIO (deliberately small)
Only simple `X is a Y` facts and `All A are B` rules are translated (grounded per named entity). Sentences with or / not / some / either and similar words are skipped, so the translation is sound but incomplete.

```python
UNSAFE = (" or ", " not ", "n't", " no ", "some ", "either", "neither", " if ", " and ",
          "only", "at least", "except", "exactly", " than ")
ENT = r"[A-Z][\w'.-]*(?: [A-Z][\w'.-]*)*"

def _parse_fact(sentence):
    m = re.match(rf"^({ENT}) (?:is|was) (?:an? |the )?(.+?)\.?$", sentence.strip())
    if m and not any(u in " " + m.group(2).lower() + " " for u in UNSAFE):
        return (norm(m.group(2)), norm(m.group(1)))        # atom = (predicate, entity)
    return None

def frontend_folio(context, statement):
    """FOLIO text -> (facts, grounded rules, no interventions, goal atom)."""
    facts, universals, entities = set(), [], set()
    for s in re.split(r"(?<=[.!?])\s+|\n", context):
        s = s.strip()
        if not s or any(u in " " + s.lower() + " " for u in UNSAFE):
            continue                                       # skip what we cannot translate soundly
        m = re.match(r"^(?:all|every|each) (.+?) (?:is|are) (?:an? )?(.+?)\.?$", s, re.I)
        if m:
            universals.append((norm(m.group(1)), norm(m.group(2))))
            continue
        f = _parse_fact(s)
        if f:
            facts.add(f); entities.add(f[1])
    goal = _parse_fact(statement)
    if goal:
        entities.add(goal[1])
    rules = [([((a, e), True)], (b, e)) for a, b in universals for e in entities]
    return facts, rules, {}, goal
```

## 4. Router, solver and evaluation
The router chooses the translator from the text (`Facts:`/`Rules:` markers → RuleTaker; "cause" in the context and "occur" in the question → CounterBench; otherwise FOLIO).
`FOLIO_DEFAULT` is the answer for FOLIO when the conclusion cannot be derived (the label `Uncertain` is treated as not entailed).

```python
FRONTENDS = {"ruletaker": frontend_ruletaker, "counterbench": frontend_counterbench, "folio": frontend_folio}

# ---- final settings ----
FOLIO_DEFAULT = False       # prediction for FOLIO when nothing is derived

def detect(context, statement=""):
    if "Facts:" in context and "Rules:" in context:
        return "ruletaker"
    if re.search(r"\bcauses?\b", context) and (statement == "" or re.search(r"\boccur", statement)):
        return "counterbench"
    return "folio"

def solve(context, statement):
    """Return True/False: is `statement` entailed by `context`?"""
    kind = detect(context, statement)
    derived = False
    try:
        facts, rules, fixed, goal = FRONTENDS[kind](context, statement)
        if goal is not None:
            known, _ = run_engine(facts, rules, fixed)
            derived = goal in known
    except Exception:
        derived = False                       # never crash on an unseen shape
    if kind == "folio" and not derived:
        return FOLIO_DEFAULT
    return derived

def rows_ruletaker(ds):    return [(r["context"], r["statement"], str(r["flag"]) == "True", r["depth"]) for r in ds]
def rows_counterbench(ds): return [(r["given_info"], r["question"], r["answer"].strip().lower() == "yes", r["type"]) for r in ds]
def rows_folio(ds):        return [(r["premises"], r["conclusion"], str(r["label"]) == "True", "all") for r in ds]

def evaluate(name, rows):
    start = time.time()
    ok, tot, wrong = Counter(), Counter(), []
    for i, (ctx, stmt, label, group) in enumerate(rows):
        good = (solve(ctx, stmt) == label)
        tot[group] += 1; ok[group] += good
        if not good:
            wrong.append(i)
    total_time = time.time() - start
    time_per_sample = total_time / len(rows)

    n_ok = sum(ok.values())
    print(f"{name}: {100 * n_ok / len(rows):.1f}% ({n_ok}/{len(rows)}) | "
          f"total {total_time:.2f}s | per sample {time_per_sample:.5f}s")
    if len(tot) > 1:
        for g in sorted(tot, key=str):
            print(f"    {g}: {100 * ok[g] / tot[g]:.1f}% ({ok[g]}/{tot[g]})")
    return wrong, time_per_sample
```

## 6. Reasoning traces
Derived atom, then the rule body that first produced it, then its supports.

```python
def show_trace(context, statement):
    kind = detect(context, statement)
    facts, rules, fixed, goal = FRONTENDS[kind](context, statement)
    known, why = run_engine(facts, rules, fixed)
    print("kind:", kind, "| goal:", goal, "| derived:", goal in known, "| fixed:", fixed)
    if goal in known:
        explain(goal, why)
```

## 7. Generate `submission.csv`
Set `TEST_PATH` to the test file (`.csv`, `.tsv`, `.json`, `.jsonl` or `.parquet`) and run this cell. It finds the columns by name
(context: `context` / `given_info` / `premises`; statement: `statement` / `question` / `conclusion`; id: `id` / `question_id` / `example_id`, otherwise the row number),
writes `id,label`, and prints the number of rows, unreadable rows (should be 0) and the total inference time.

```python
import os, time, csv
import pandas as pd

# ================== the only line to change when the test file arrives ==================
TEST_PATH = "/content/private_test.csv"            # .csv / .tsv / .json / .jsonl / .parquet
OUT_PATH  = "submission.csv"
# =========================================================================================

# ---- final settings (reset anything the experiments changed) ----
NEG_MODE = "naf"
NESTED_MODE = "both"
FOLIO_DEFAULT = False             # prediction for FOLIO when nothing is derived

# ---- reading the test file ----
CTX_COLS  = ["context", "given_info", "premises", "knowledge_base", "kb", "text"]
STMT_COLS = ["statement", "question", "conclusion", "query", "hypothesis"]
ID_COLS   = ["id", "question_id", "example_id", "idx", "index"]

def _missing(v):
    return v is None or (isinstance(v, float) and v != v)

def _pick(row, names):
    for n in names:
        v = row.get(n)
        if not _missing(v):
            return v
    return None

def load_test(path):
    if path.endswith(".jsonl"):
        df = pd.read_json(path, lines=True)
    elif path.endswith(".json"):
        df = pd.read_json(path)
    elif path.endswith(".parquet"):
        df = pd.read_parquet(path)
    elif path.endswith(".tsv"):
        df = pd.read_csv(path, sep="\t")
    else:
        df = pd.read_csv(path)
    df.columns = [str(c).strip().lower() for c in df.columns]
    return df.to_dict("records")

def make_submission(rows, path=OUT_PATH):
    start, unreadable = time.time(), 0
    with open(path, "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["id", "label"])
        for i, row in enumerate(rows):
            ctx, stmt = _pick(row, CTX_COLS), _pick(row, STMT_COLS)
            rid = _pick(row, ID_COLS)
            rid = i if rid is None else rid
            if ctx is None or stmt is None:
                unreadable += 1
                pred = FOLIO_DEFAULT
            else:
                pred = solve(str(ctx), str(stmt))
            w.writerow([rid, pred])
    print(f"wrote {path} | rows: {len(rows)} | unreadable: {unreadable} | inference time: {time.time() - start:.2f}s")
    return unreadable

# ---- run ----
if not os.path.exists(TEST_PATH):
    print(f"Test file '{TEST_PATH}' not found. Set TEST_PATH above and run this cell again.")
else:
    test_rows = load_test(TEST_PATH)
    print(len(test_rows), "rows | columns:", list(test_rows[0].keys()))
    bad = make_submission(test_rows)
    if bad:
        print("Some rows had no recognised context/statement column. Columns found above;",
              "add the right names to CTX_COLS / STMT_COLS.")
    sub = pd.read_csv(OUT_PATH)
    print(sub["label"].value_counts().to_string())
    print("ids unique:", sub["id"].is_unique)
truth = pd.read_csv(TEST_PATH)
sub = pd.read_csv(OUT_PATH)


# make sure both labels are True/False, not strings
truth["label"] = truth["label"].astype(str).str.strip().str.lower() == "true"
sub["label"] = sub["label"].astype(str).str.strip().str.lower() == "true"


correct = (truth["label"] == sub["label"]).sum()
print(f"accuracy: {correct}/{len(truth)} = {correct / len(truth):.2%}")
```

```python
# ===== Improvements found by testing on private_test.csv =====
# 1) CounterBench: the sentence 'Nothing else causes any of these events.' was parsed as a rule, and
#    'X did not occur' observations were parsed wrongly.  2) RuleTaker written as plain sentences.
# 3) First-order logic entailment (for FOLIO-style rows that come with a logic form).
import re


def _parse_observed(text):
    """'Drav. Vapo did not occur.' / 'that Fesh and Vapo did not occur' -> {name: True/False}"""
    observed = {}
    for sent in re.split(r"(?<=\.)\s+", text.strip()):
        sent = re.sub(r"^that\s+", "", sent.strip().rstrip(".")).strip()
        if not sent:
            continue
        positive = True
        if re.search(r"did not occur$|do not occur$|did not happen$", sent):
            positive = False
            sent = re.sub(r"\s*(did|do) not (occur|happen)$", "", sent)
        for part in re.split(r",\s*|\s+and\s+", sent):
            part = part.strip()
            if part:
                name, val = _lit(part)            # also understands "not X" (public CounterBench wording)
                observed[name] = val and positive
    return observed


def frontend_counterbench(context, statement):
    """CounterBench text -> (root facts, causal rules, interventions, target atom)."""
    text, observed = context.strip(), {}
    text = re.sub(r"Nothing else causes any of these events\.?", "", text)   # closed-world marker, not a rule
    m = re.search(r"We observed (.+)$", text, re.S)
    if m:
        observed = _parse_observed(m.group(1))
        text = text[:m.start()]
    text = re.sub(r"^\sWe know that\s", "", text).strip().rstrip(". ")


    segments = re.split(r"\bcauses?\b", text)
    bodies, spec = [_body_dnf(segments[0])], []
    for k in range(1, len(segments)):
        pieces = [p.strip() for p in segments[k].split(",") if p.strip()]
        last = (k == len(segments) - 1)
        eff_pieces = pieces if last else pieces[:-1]
        effects = []
        for p in eff_pieces:
            for q in re.sub(r"^and\s+", "", p).split(" and "):
                effects.append(_lit(q))
        spec.append((bodies[-1], effects))
        if not last:
            bodies.append(_body_dnf(pieces[-1]))


    rules, heads, used = [], set(), set()
    for dnf, effects in spec:
        for name, positive in effects:
            for conj in (dnf if positive else _negate_dnf(dnf)):
                rules.append((conj, name))
                heads.add(name)
                used.update(a for a, _ in conj)
    roots = (used - heads) | {n for n in observed if n not in heads}
    facts = {r for r in roots if observed.get(r, True)}


    q = statement.strip()
    mt = re.search(r"[Ww]ould\s+(\w+)\s+occur", q)
    if not mt:
        return facts, rules, {}, None
    target = mt.group(1)
    cond = re.split(r"\binstead of\b", q[:mt.start()] + " " + q[mt.end():])[0]
    atoms = heads | used | {target}
    ordered = [(m2.group(2), m2.group(1) is None)
               for m2 in re.finditer(r"(not\s+)?\b([A-Z]\w*)\b", cond) if m2.group(2) in atoms]
    if re.search(r"further suppose|based on this assumption", q, re.I) and len(ordered) > 1:
        if NESTED_MODE == "first_only":
            ordered = ordered[:1]
        elif NESTED_MODE == "last_only":
            ordered = ordered[-1:]
    return facts, rules, dict(ordered), target




# ---------- RuleTaker, plain-sentence version ----------


VAR_WORDS = ("someone", "something", "they", "it")


def _ent(text):
    text = text.strip().lower()
    return re.sub(r"^the\s+", "", text)


def _subject(text):
    t = _ent(text)
    return "X" if t in VAR_WORDS else t


def _literal(text):
    """'someone is cold' / 'they help yarra' / 'the otter is not quiet' -> (neg, pred, subj, obj)"""
    text = text.strip().lower()
    m = re.match(r"^(.+?) (?:is|are) (not )?([a-z]+)$", text)
    if m:
        return (bool(m.group(2)), m.group(3), _subject(m.group(1)), None)
    m = re.match(r"^(.+?) (?:does not|do not) ([a-z]+) (.+)$", text)
    if m:
        return (True, m.group(2), _subject(m.group(1)), _ent(m.group(3)))
    m = re.match(r"^(\w+(?: \w+)??) ([a-z]+?)s? (?:the )?([a-z]+)$", text)   # 'yarra visits the otter' / 'they help yarra'
    if m:
        verb = m.group(2)
        return (False, verb, _subject(m.group(1)), _ent(m.group(3)))
    return None


def parse_rt_nl(context):
    """Return (facts, rules) or None if any sentence is not understood."""
    facts, rules = set(), []
    for sent in re.split(r"(?<=\.)\s+", context.strip()):
        s = sent.strip().rstrip(".").strip()
        if not s:
            continue
        low = s.lower()
        m = re.match(r"^if (.+?) then (.+)$", low)
        if m:
            body = [_literal(p) for p in re.split(r" and ", m.group(1))]
            head = _literal(m.group(2))
            if None in body or head is None or head[0]:
                return None
            rules.append((body, head))
            continue
        m = re.match(r"^(?:all )?(.+?) things are ([a-z]+)$", low)
        if m:
            body = [(False, a.strip(), "X", None) for a in m.group(1).split(",")]
            rules.append((body, (False, m.group(2), "X", None)))
            continue
        lit = _literal(low)
        if lit is None or lit[0] or lit[2] == "X":
            return None
        facts.add(lit[1:])
    return facts, rules


def solve_rt_nl(context, statement):
    parsed = parse_rt_nl(context)
    if parsed is None:
        return None
    facts, rules = parsed
    goal = _literal(statement.strip().rstrip(".").lower())
    if goal is None:
        return None
    ents = {f[1] for f in facts} | {f[2] for f in facts if f[2]} | {goal[2]}
    for body, head in rules:
        for lit in body + [head]:
            if lit[2] != "X": ents.add(lit[2])
            if lit[3]: ents.add(lit[3])
    known = set(facts)
    changed = True
    while changed:
        changed = False
        for body, head in rules:
            for e in ents:
                def ground(l): return (l[1], e if l[2] == "X" else l[2], l[3])
                if all((ground(l) in known) != l[0] for l in body):
                    h = ground(head)
                    if h not in known:
                        known.add(h); changed = True
    neg, pred, subj, obj = goal
    inside = (pred, subj, obj) in known
    return (not inside) if neg else inside


# ---------- first-order logic: parser + SAT ----------


# ---------- parser:  text -> formula tree ----------
TOKEN = re.compile(r"∀|∃|¬|∧|∨|⊕|→|↔|\(|\)|,|[A-Za-z_][A-Za-z0-9_]*")


class Parser:
    def __init__(self, text):
        self.t = TOKEN.findall(text); self.i = 0
    def peek(self): return self.t[self.i] if self.i < len(self.t) else None
    def eat(self, x=None):
        tok = self.peek()
        if x is not None and tok != x: raise ValueError("expected %s got %s" % (x, tok))
        self.i += 1; return tok
    def iff(self):
        a = self.imp()
        while self.peek() == "↔": self.eat(); a = ("iff", a, self.imp())
        return a
    def imp(self):
        a = self.orr()
        if self.peek() == "→": self.eat(); return ("imp", a, self.imp())
        return a
    def orr(self):
        a = self.xor()
        while self.peek() == "∨": self.eat(); a = ("or", a, self.xor())
        return a
    def xor(self):
        a = self.andd()
        while self.peek() == "⊕": self.eat(); a = ("xor", a, self.andd())
        return a
    def andd(self):
        a = self.unary()
        while self.peek() == "∧": self.eat(); a = ("and", a, self.unary())
        return a
    def unary(self):
        tok = self.peek()
        if tok == "¬": self.eat(); return ("not", self.unary())
        if tok in ("∀", "∃"):
            self.eat(); var = self.eat()
            return ("all" if tok == "∀" else "ex", var, self.unary() if self.peek() != "(" else self.unary())
        if tok == "(":
            self.eat(); f = self.iff(); self.eat(")"); return f
        name = self.eat(); args = []
        if self.peek() == "(":
            self.eat()
            while True:
                args.append(self.eat())
                if self.peek() == ",": self.eat(); continue
                break
            self.eat(")")
        return ("atom", name, tuple(args))


def parse_fol(text):
    p = Parser(text); f = p.iff()
    if p.peek() is not None: raise ValueError("trailing tokens")
    return f


# ---------- helpers ----------
def fol_constants(f, bound=()):
    k = f[0]
    if k == "atom": return {a for a in f[2] if a not in bound}
    if k in ("all", "ex"): return fol_constants(f[2], bound + (f[1],))
    out = set()
    for sub in f[1:]: out |= fol_constants(sub, bound)
    return out


def count_ex(f, pos=True):
    k = f[0]
    if k == "atom": return 0
    if k == "not": return count_ex(f[1], not pos)
    if k == "ex":  return (1 if pos else 0) + count_ex(f[2], pos)
    if k == "all": return (0 if pos else 1) + count_ex(f[2], pos)
    if k == "imp": return count_ex(f[1], not pos) + count_ex(f[2], pos)
    if k in ("xor", "iff"): return sum(count_ex(s, p) for s in f[1:] for p in (True, False))
    return count_ex(f[1], pos) + count_ex(f[2], pos)


# ---------- grounding + Tseitin CNF ----------
class Encoder:
    def __init__(self, domain):
        self.domain, self.n, self.clauses, self.atoms, self.cache = domain, 1, [[1]], {}, {}   # var 1 is TRUE
    def new(self): self.n += 1; return self.n
    def atom(self, key):
        if key not in self.atoms: self.atoms[key] = self.new()
        return self.atoms[key]
    def enc(self, f, env):
        k = f[0]
        if k == "atom":
            return self.atom((f[1], tuple(env.get(a, a) for a in f[2])))
        if k == "not": return -self.enc(f[1], env)
        if k in ("and", "or", "imp", "xor", "iff"):
            x, y = self.enc(f[1], env), self.enc(f[2], env)
            if k == "imp": k, x = "or", -x
            key = (k,) + tuple(sorted((x, y)))
            if key in self.cache: return self.cache[key]
            g = self.new(); C = self.clauses; self.cache[key] = g
            if k == "and": C += [[-g, x], [-g, y], [g, -x, -y]]
            elif k == "or": C += [[-g, x, y], [g, -x], [g, -y]]
            elif k == "xor": C += [[-g, x, y], [-g, -x, -y], [g, -x, y], [g, x, -y]]
            else: C += [[-g, x, -y], [-g, -x, y], [g, x, y], [g, -x, -y]]
            return g
        kids = [self.enc(f[2], {**env, f[1]: d}) for d in self.domain]
        key = (k,) + tuple(sorted(set(kids)))
        if key in self.cache: return self.cache[key]
        g = self.new(); C = self.clauses; self.cache[key] = g
        if k == "all":
            C += [[-g, x] for x in kids] + [[g] + [-x for x in kids]]
        else:
            C += [[-g] + kids] + [[g, -x] for x in kids]
        return g
    def assert_true(self, f, env=None):
        env = env or {}; k = f[0]
        if k == "and": self.assert_true(f[1], env); self.assert_true(f[2], env)
        elif k == "all":
            for d in self.domain: self.assert_true(f[2], {**env, f[1]: d})
        else: self.clauses.append([self.enc(f, env)])


# ---------- DPLL with unit propagation ----------
def dpll_sat(clauses, nvars, limit=300000):
    watch = {}
    for ci, c in enumerate(clauses):
        for l in c[:2]: watch.setdefault(l, []).append(ci)
    assign, trail, steps = {}, [], [0]
    def val(l):
        v = assign.get(abs(l)); return None if v is None else (v if l > 0 else not v)
    def set_lit(l, queue):
        assign[abs(l)] = l > 0; trail.append(abs(l)); queue.append(l)
    def propagate(queue):
        while queue:
            l = queue.pop(); false_lit = -l
            for ci in list(watch.get(false_lit, [])):
                c = clauses[ci]
                if c[0] == false_lit: c[0], c[1] = c[1], c[0]
                if len(c) > 1 and val(c[0]) is True: continue
                moved = False
                for j in range(2, len(c)):
                    if val(c[j]) is not False:
                        c[1], c[j] = c[j], c[1]
                        watch[false_lit].remove(ci); watch.setdefault(c[1], []).append(ci)
                        moved = True; break
                if moved: continue
                if len(c) == 1 or val(c[0]) is False: return False
                if val(c[0]) is None: set_lit(c[0], queue)
        return True
    q = []
    for c in clauses:
        if len(c) == 1:
            if val(c[0]) is False: return False
            if val(c[0]) is None: set_lit(c[0], q)
    if not propagate(q): return False
    order = list(range(2, nvars + 1))
    def solve_rec():
        steps[0] += 1
        if steps[0] > limit: raise TimeoutError
        v = next((x for x in order if x not in assign), None)
        if v is None: return True
        for choice in (v, -v):
            mark, q = len(trail), []
            set_lit(choice, q)
            if propagate(q) and solve_rec(): return True
            while len(trail) > mark: del assign[trail.pop()]
        return False
    return solve_rec()


# ---------- public entry ----------
def entails(context_logic, statement_logic):
    """True iff premises |= conclusion  (checked as: premises AND NOT conclusion is unsatisfiable)."""
    premises = [parse_fol(l.strip()) for l in context_logic.split("\n") if l.strip()]
    concl = parse_fol(statement_logic.strip())
    neg = ("not", concl)
    consts = set()
    for f in premises + [neg]: consts |= fol_constants(f)
    extra = sum(count_ex(f) for f in premises + [neg])
    domain = sorted(consts) + ["_w%d" % i for i in range(extra)]
    if not domain: domain = ["_w0"]
    enc = Encoder(domain)
    for f in premises + [neg]: enc.assert_true(f)
    try:
        return not dpll_sat(enc.clauses, enc.n)
    except TimeoutError:
        return False


# ======================= router: pick the right translator =======================
FRONTENDS = {"ruletaker": frontend_ruletaker, "counterbench": frontend_counterbench, "folio": frontend_folio}


def solve(context, statement, ctx_logic=None, stmt_logic=None):
    """Return True/False: is statement entailed by context?
    1) Facts:/Rules: format -> RuleTaker engine      2) 'causes' + 'occur' -> CounterBench engine
    3) plain 'someone/things' rules -> RuleTaker-NL engine
    4) otherwise, if a logic form is supplied -> first-order entailment (SAT)
    5) last resort -> the small FOLIO translator"""
    kind = detect(context, statement)
    if kind in ("ruletaker", "counterbench"):
        try:
            facts, rules, fixed, goal = FRONTENDS[kind](context, statement)
            if goal is None:
                return False
            known, _ = run_engine(facts, rules, fixed)
            return goal in known
        except Exception:
            return False
    try:
        answer = solve_rt_nl(context, statement)
        if answer is not None:
            return answer
    except Exception:
        pass
    if ctx_logic and stmt_logic and any(s in str(ctx_logic) for s in "\u2200\u2203\u2227\u2228\u2192\u00ac"):
        try:
            return entails(str(ctx_logic), str(stmt_logic))
        except Exception:
            pass
    derived = False
    try:
        facts, rules, fixed, goal = frontend_folio(context, statement)
        if goal is not None:
            known, _ = run_engine(facts, rules, fixed)
            derived = goal in known
    except Exception:
        pass
    return derived if derived else FOLIO_DEFAULT
```
